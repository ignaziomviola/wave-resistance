# Michell wave resistance of a Wigley hull

This example evaluates the parameter-free thin-ship model over $Fn=0.10$ to $0.45$. The hull is fixed at its static waterline in calm, effectively infinite-depth water.

```python
import numpy as np
import matplotlib.pyplot as plt
from wave_resistance import MichellSolver, wigley_hull
```

```python
hull = wigley_hull(length_m=4.0, nx=81, nz=33)
fn = np.arange(0.10, 0.4501, 0.005)
result = MichellSolver().solve(hull, fn)
assert result.converged.all()
```

```python
fig, ax = plt.subplots(figsize=(7, 4.2), constrained_layout=True)
ax.plot(result.fn, 1e3 * result.c_wave_resistance, lw=2)
ax.set(xlabel=r'$Fn=U/\sqrt{gL}$', ylabel=r'$10^3 C_{R_w}$')
ax.grid(alpha=0.25)
plt.show()
```

Always inspect `result.tail_fraction`, `result.cancellation_ratio`, and `result.validity_flags` together with the resistance curve. Numerical convergence does not remove Michell theory's model-form limitations.
